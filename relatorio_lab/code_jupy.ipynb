```python
import random
import matplotlib.pyplot as plt

def simula_tcp_reno(rtts=30, ssthresh_inicial=16, prob_perda=0.1, seed=1):
    random.seed(seed)
    cwnd, ssthresh = 1.0, ssthresh_inicial
    historico = []
    for rtt in range(1, rtts + 1):
        if random.random() < prob_perda:
            ssthresh = max(cwnd / 2, 1)
            cwnd = ssthresh + 3          # Fast Recovery (Reno)
        elif cwnd < ssthresh:
            cwnd *= 2                     # Slow Start
        else:
            cwnd += 1                     # Congestion Avoidance (AIMD)
        historico.append((rtt, cwnd, ssthresh))
    return historico

dados = simula_tcp_reno()
rtts = [d[0] for d in dados]
cwnds = [d[1] for d in dados]

plt.plot(rtts, cwnds, marker="o")
plt.xlabel("RTT"); plt.ylabel("cwnd (MSS)")
plt.title("Simulação TCP Reno: Slow Start + AIMD + Fast Recovery")
plt.grid(True)
plt.savefig("tcp_reno.png", dpi=150)
plt.show()
```
